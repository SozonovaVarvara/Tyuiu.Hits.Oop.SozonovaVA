[Создать базовый класс Ticket в C#, который будет представлять билеты на различные мероприятия. На основе этого класса разработать 2-3 производных класса, демонстрирующих принципы наследования и полиморфизма. В каждом из классов должны быть реализованы новые атрибуты и методы, а также переопределены некоторые методы базового класса для демонстрации полиморфизма.

Требования к базовому классу Ticket:
    • Атрибуты: ID билета (TicketId), Номер ряда (RowNumber), Номер места (SeatNumber). 
    • Методы:  
        o PrintTicketInfo(): метод для печати информации о билете.
        o ReserveSeat(): метод для резервирования места. 
        o GetTicketDetails(): метод для получения деталей билета.

Требования к производным классам:
1) БилетНаКонцерт (ConcertTicket): Должен содержать дополнительные атрибуты, такие как Имя исполнителя (ArtistName). Метод PrintTicketInfo() должен быть переопределен для включения имени исполнителя в информацию о билете.
2) БилетНаСпектакль (PlayTicket): Должен содержать дополнительные атрибуты, такие как Название спектакля (PlayTitle). Метод GetTicketDetails() должен быть переопределен для отображения названия спектакля вместе с другими деталями билета.
3) БилетНаКонференцию (ConferenceTicket) (если требуется третий класс): Должен содержать дополнительные атрибуты, такие как Название конференции (ConferenceName). Метод ReserveSeat() должен быть переопределен для добавления информации о предпочтениях участника относительно мест.]

#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) и реализуйте простое, сложное и множественное наследование

```
using System;
using System.Collections.Generic;

// ИНТЕРФЕЙС ДЛЯ МНОЖЕСТВЕННОГО НАСЛЕДОВАНИЯ

// Интерфейс онлайн-билетов
public interface IOnlineTicket
{
    string Platform { get; set; }
    void SendToEmail(string email);
}

// Интерфейс скидочного билета
public interface IDiscountTicket
{
    double Discount { get; set; }
    double GetPriceWithDiscount();
}

// БАЗОВЫЙ КЛАСС TICKET

public class Ticket
{
    // Основные свойства
    public int TicketId { get; set; }
    public int RowNumber { get; set; }
    public int SeatNumber { get; set; }
    public bool IsReserved { get; private set; }

    // ДОБАВЛЕНЫЕ АТРИБУТЫ
    public double Price { get; set; }
    public string Date { get; set; }
    public string Hall { get; set; }
    public string BuyerName { get; set; }

    // Конструктор
    public Ticket(int ticketId, int rowNumber, int seatNumber)
    {
        TicketId = ticketId;
        RowNumber = rowNumber;
        SeatNumber = seatNumber;
        IsReserved = false;

        Price = 0;
        Date = "";
        Hall = "";
        BuyerName = "";
    }

    // Вывод информации
    public virtual void PrintTicketInfo()
    {
        Console.WriteLine(
            $"Билет #{TicketId}: ряд {RowNumber}, место {SeatNumber}");
    }

    // Бронь места
    public virtual void ReserveSeat()
    {
        if (!IsReserved)
        {
            IsReserved = true;

            Console.WriteLine(
                $"Место {SeatNumber} в ряду {RowNumber} успешно забронировано!");
        }
        else
        {
            Console.WriteLine("Это место уже забронировано.");
        }
    }

    // Получение полной информации
    public virtual string GetTicketDetails()
    {
        return $"Ticket ID: {TicketId}, Row: {RowNumber}, " +
               $"Seat: {SeatNumber}, Reserved: {IsReserved}, " +
               $"Price: {Price}, Date: {Date}, Hall: {Hall}, " +
               $"Buyer: {BuyerName}";
    }

    // ДОБАВЛЕННЫЙ МЕТОД 1
    public virtual void CancelReservation()
    {
        if (IsReserved)
        {
            IsReserved = false;
            Console.WriteLine($"Бронирование билета #{TicketId} отменено.");
        }
        else
        {
            Console.WriteLine("Билет не был забронирован.");
        }
    }

    // ДОБАВЛЕННЫЙ МЕТОД 2
    public void ChangeSeat(int newRow, int newSeat)
    {
        RowNumber = newRow;
        SeatNumber = newSeat;

        Console.WriteLine(
            $"Место изменено: ряд {RowNumber}, место {SeatNumber}");
    }

    // ДОБАВЛЕННЫЙ МЕТОД 3
    public void SetBuyer(string name)
    {
        BuyerName = name;
        Console.WriteLine($"Покупатель: {BuyerName}");
    }

    // ДОБАВЛЕННЫЙ МЕТОД 4
    public virtual void ShowPrice()
    {
        Console.WriteLine($"Цена билета: {Price} руб.");
    }
}

// ======================================================
// ПРОИЗВОДНЫЙ КЛАСС 1 - CONCERT TICKET
// ПРОСТОЕ НАСЛЕДОВАНИЕ
// ConcertTicket -> Ticket
// ======================================================

public class ConcertTicket : Ticket
{
    // Доп свойства
    public string ArtistName { get; set; }
    public string Genre { get; set; }
    public string Stage { get; set; }
    public bool MeetAndGreet { get; set; }

    // Конструктор
    public ConcertTicket(
        int ticketId,
        int rowNumber,
        int seatNumber,
        string artistName)
        : base(ticketId, rowNumber, seatNumber)
    {
        ArtistName = artistName;
        Genre = "";
        Stage = "";
        MeetAndGreet = false;
    }

    // Переопределение метода
    public override void PrintTicketInfo()
    {
        Console.WriteLine(
            $"[Концерт] Билет #{TicketId}: {ArtistName}, " +
            $"ряд {RowNumber}, место {SeatNumber}");
    }

    // Доп метод 1
    public void ShowArtist()
    {
        Console.WriteLine($"Исполнитель: {ArtistName}");
    }

    // Доп метод 2
    public void ChangeStage(string newStage)
    {
        Stage = newStage;
        Console.WriteLine($"Сцена изменена на: {Stage}");
    }

    // Доп метод 3
    public void AddMeetAndGreet()
    {
        MeetAndGreet = true;
        Console.WriteLine("В билет добавлена встреча с артистом.");
    }

    // Доп метод 4
    public override void ShowPrice()
    {
        Console.WriteLine(
            $"Цена концертного билета: {Price} руб.");
    }
}

// ======================================================
// ПРОИЗВОДНЫЙ КЛАСС 2 - PLAY TICKET
// ======================================================

public class PlayTicket : Ticket
{
    public string PlayTitle { get; set; }
    public string Director { get; set; }
    public string TheaterName { get; set; }
    public string Language { get; set; }

    // Конструктор
    public PlayTicket(
        int ticketId,
        int rowNumber,
        int seatNumber,
        string playTitle)
        : base(ticketId, rowNumber, seatNumber)
    {
        PlayTitle = playTitle;
        Director = "";
        TheaterName = "";
        Language = "Русский";
    }

    // Переопределение метода
    public override string GetTicketDetails()
    {
        return $"[Спектакль] {PlayTitle} - " +
               $"Ticket ID: {TicketId}, Row: {RowNumber}, " +
               $"Seat: {SeatNumber}, Reserved: {IsReserved}, " +
               $"Director: {Director}, Theater: {TheaterName}";
    }

    // Доп метод 1
    public void ShowPlay()
    {
        Console.WriteLine($"Название спектакля: {PlayTitle}");
    }

    // Доп метод 2
    public void ChangeDirector(string director)
    {
        Director = director;
        Console.WriteLine($"Режиссер: {Director}");
    }

    // Доп метод 3
    public void ChangeLanguage(string language)
    {
        Language = language;
        Console.WriteLine($"Язык спектакля: {Language}");
    }

    // Доп метод 4
    public void ShowTheater()
    {
        Console.WriteLine($"Театр: {TheaterName}");
    }
}

// ======================================================
// ПРОИЗВОДНЫЙ КЛАСС 3 - CONFERENCE TICKET
// ======================================================

public class ConferenceTicket : Ticket
{
    public string ConferenceName { get; set; }
    public string SeatPreference { get; set; }
    public string Speaker { get; set; }
    public string Topic { get; set; }

    // Конструктор
    public ConferenceTicket(
        int ticketId,
        int rowNumber,
        int seatNumber,
        string conferenceName)
        : base(ticketId, rowNumber, seatNumber)
    {
        ConferenceName = conferenceName;
        SeatPreference = "";
        Speaker = "";
        Topic = "";
    }

    // Переопределение бронирования
    public override void ReserveSeat()
    {
        if (!string.IsNullOrEmpty(SeatPreference))
        {
            Console.WriteLine(
                $"Предпочтение: {SeatPreference}");
        }

        base.ReserveSeat();
    }

    //  Доп метод 1
    public void SetSpeaker(string speaker)
    {
        Speaker = speaker;
        Console.WriteLine($"Докладчик: {Speaker}");
    }

    //  Доп метод 2
    public void SetTopic(string topic)
    {
        Topic = topic;
        Console.WriteLine($"Тема доклада: {Topic}");
    }

    // Доп метод 3
    public void ShowConference()
    {
        Console.WriteLine($"Конференция: {ConferenceName}");
    }

    //  Доп метод 4
    public void ShowPreference()
    {
        Console.WriteLine($"Предпочтение места: {SeatPreference}");
    }
}

// ======================================================
// СЛОЖНОЕ НАСЛЕДОВАНИЕ
// 1 - Ticket
// 2 - ConcertTicket
// 3 - VipConcertTicket
// ======================================================

public class VipConcertTicket : ConcertTicket
{
    public string VipZone { get; set; }
    public bool BackstageAccess { get; set; }
    public string WelcomeDrink { get; set; }
    public string Gift { get; set; }

    // Конструктор
    public VipConcertTicket(
        int ticketId,
        int rowNumber,
        int seatNumber,
        string artistName)
        : base(ticketId, rowNumber, seatNumber, artistName)
    {
        VipZone = "VIP";
        BackstageAccess = false;
        WelcomeDrink = "";
        Gift = "";
    }

    // Переопределение метода
    public override void PrintTicketInfo()
    {
        Console.WriteLine(
            $"[VIP КОНЦЕРТ] Билет #{TicketId}: {ArtistName}, " +
            $"ряд {RowNumber}, место {SeatNumber}, зона {VipZone}");
    }

    //  Доп метод 1
    public void AddBackstageAccess()
    {
        BackstageAccess = true;
        Console.WriteLine("Добавлен доступ за кулисы.");
    }

    // Доп метод 2
    public void SetWelcomeDrink(string drink)
    {
        WelcomeDrink = drink;
        Console.WriteLine($"Приветственный напиток: {WelcomeDrink}");
    }

    // Доп метод 3
    public void SetGift(string gift)
    {
        Gift = gift;
        Console.WriteLine($"VIP-подарок: {Gift}");
    }

    // Доп метод 4
    public void ShowVipInfo()
    {
        Console.WriteLine(
            $"VIP-зона: {VipZone}, Доступ за кулисы: {BackstageAccess}");
    }
}

// ======================================================
// МНОЖЕСТВЕННОЕ НАСЛЕДОВАНИЕ ЧЕРЕЗ ИНТЕРФЕЙСЫ
// OnlineTicket наследуется от Ticket
// и реализует сразу 2 интерфейса:
// IOnlineTicket + IDiscountTicket
// ======================================================

public class OnlineTicket : Ticket, IOnlineTicket, IDiscountTicket
{
    public string Platform { get; set; }
    public double Discount { get; set; }

    public string QRCode { get; set; }
    public string Email { get; set; }

    // Конструктор
    public OnlineTicket(
        int ticketId,
        int rowNumber,
        int seatNumber)
        : base(ticketId, rowNumber, seatNumber)
    {
        Platform = "";
        Discount = 0;
        QRCode = "";
        Email = "";
    }

    // Реализация метода IOnlineTicket
    public void SendToEmail(string email)
    {
        Email = email;

        Console.WriteLine(
            $"Билет #{TicketId} отправлен на почту {Email}");
    }

    // Реализация метода IDiscountTicket
    public double GetPriceWithDiscount()
    {
        return Price - (Price * Discount / 100);
    }

    //  Доп метод
    public void GenerateQRCode()
    {
        QRCode = "QR-" + TicketId;
        Console.WriteLine($"QR-код создан: {QRCode}");
    }

    // Переопределение базвого метода
    public override void PrintTicketInfo()
    {
        Console.WriteLine(
            $"[ОНЛАЙН] Билет #{TicketId}, платформа: {Platform}, " +
            $"цена: {Price} руб.");
    }
}

// ======================================================
// ПЕРЕГРУЗКА МЕТОДА
// ======================================================

public class TicketHelper
{
    // Метод с одним параметром
    public void PrintTicket(string name)
    {
        Console.WriteLine($"Билет принадлежит: {name}");
    }

    // Перегрузка с двумя параметрами
    public void PrintTicket(string name, int ticketId)
    {
        Console.WriteLine(
            $"Билет #{ticketId} принадлежит: {name}");
    }

    // Перегрузка с тремя параметрами
    public void PrintTicket(string name, int ticketId, double price)
    {
        Console.WriteLine(
            $"Билет #{ticketId}: {name}, цена {price} руб.");
    }
}

// ======================================================
// КЛАСС TICKETSYSTEM
// ======================================================

public class TicketSystem
{
    public List<Ticket> Tickets { get; set; }

    // Доп свойства
    public string SystemName { get; set; }
    public string City { get; set; }
    public int MaxTickets { get; set; }
    public bool IsOpen { get; set; }

    // Конструктор
    public TicketSystem()
    {
        Tickets = new List<Ticket>();

        SystemName = "Ticket System";
        City = "";
        MaxTickets = 1000;
        IsOpen = true;
    }

    // Добавление билета
    public void AddTicket(Ticket ticket)
    {
        if (Tickets.Count < MaxTickets)
        {
            Tickets.Add(ticket);

            Console.WriteLine(
                $"Билет #{ticket.TicketId} добавлен в систему.");
        }
        else
        {
            Console.WriteLine("Достигнут лимит билетов.");
        }
    }

    // Вывод всех билетов
    public void ShowAllTickets()
    {
        Console.WriteLine("\n--- Все билеты ---");

        foreach (Ticket ticket in Tickets)
        {
            ticket.PrintTicketInfo();
        }
    }

    // Бронирование билета
    public void ReserveTicket(int ticketId)
    {
        foreach (Ticket ticket in Tickets)
        {
            if (ticket.TicketId == ticketId)
            {
                ticket.ReserveSeat();
                return;
            }
        }

        Console.WriteLine("Билет с таким ID не найден.");
    }

    // Доп метод
    public void RemoveTicket(int ticketId)
    {
        Ticket ticketToRemove = null;

        foreach (Ticket ticket in Tickets)
        {
            if (ticket.TicketId == ticketId)
            {
                ticketToRemove = ticket;
                break;
            }
        }

        if (ticketToRemove != null)
        {
            Tickets.Remove(ticketToRemove);
            Console.WriteLine($"Билет #{ticketId} удален.");
        }
        else
        {
            Console.WriteLine("Билет не найден.");
        }
    }

    // Дополнительный метод
    public int GetTicketCount()
    {
        return Tickets.Count;
    }

    // Доп метод
    public void ShowSystemInfo()
    {
        Console.WriteLine(
            $"Система: {SystemName}, город: {City}, " +
            $"билетов: {Tickets.Count}");
    }

    // Доп метод
    public void ClearTickets()
    {
        Tickets.Clear();
        Console.WriteLine("Все билеты удалены.");
    }
}

// ======================================================
// РАБОТА ПРОГРАММЫ
// ======================================================

Console.WriteLine("=================================");
Console.WriteLine("       СИСТЕМА БИЛЕТОВ");
Console.WriteLine("=================================");

// ------------------------------------------------------
// ПРОСТОЕ НАСЛЕДОВАНИЕ
// ------------------------------------------------------

Console.WriteLine("\n--- Простое наследование ---");

Ticket concertTicket =
    new ConcertTicket(1, 5, 34, "Kanye West");

ConcertTicket concert =
    new ConcertTicket(2, 3, 15, "The Weeknd");

concert.Genre = "Pop";
concert.Stage = "Main Stage";
concert.Price = 5000;

concert.PrintTicketInfo();
concert.ShowArtist();
concert.ChangeStage("Big Stage");
concert.ShowPrice();

// ------------------------------------------------------
// PLAY TICKET
// ------------------------------------------------------

Console.WriteLine("\n--- Билет на спектакль ---");

PlayTicket playTicket =
    new PlayTicket(3, 2, 13, "Лебединое озеро");

playTicket.Price = 3500;
playTicket.Director = "Алексей Иванов";
playTicket.TheaterName = "Большой театр";

playTicket.PrintTicketInfo();
playTicket.ShowPlay();
playTicket.ShowTheater();

Console.WriteLine(playTicket.GetTicketDetails());

// ------------------------------------------------------
// CONFERENCE TICKET
// ------------------------------------------------------

Console.WriteLine("\n--- Билет на конференцию ---");

ConferenceTicket conferenceTicket =
    new ConferenceTicket(
        4,
        7,
        15,
        "TechConf 2026");

conferenceTicket.Price = 2500;
conferenceTicket.SeatPreference = "Ближе к сцене";

conferenceTicket.SetSpeaker("Иван Петров");
conferenceTicket.SetTopic("Искуственный интеллект");
conferenceTicket.ShowConference();
conferenceTicket.ShowPreference();

conferenceTicket.ReserveSeat();

// ------------------------------------------------------
// МНОГОУРОВНЕВОЕ НАСЛЕДОВАНИЕ
// ------------------------------------------------------

Console.WriteLine("\n--- Многоуровневое наследование ---");

VipConcertTicket vipTicket =
    new VipConcertTicket(
        5,
        1,
        1,
        "Kanye West");

vipTicket.Price = 15000;
vipTicket.VipZone = "VIP GOLD";

vipTicket.PrintTicketInfo();
vipTicket.AddBackstageAccess();
vipTicket.SetWelcomeDrink("Шампанское");
vipTicket.SetGift("VIP Box");
vipTicket.ShowVipInfo();

// ------------------------------------------------------
// МНОЖЕСТВЕННОЕ НАСЛЕДОВАНИЕ
// ------------------------------------------------------

Console.WriteLine("\n--- Множественное наследование через интерфейсы ---");

OnlineTicket onlineTicket =
    new OnlineTicket(6, 0, 0);

onlineTicket.Price = 4000;
onlineTicket.Platform = "TicketOnline";
onlineTicket.Discount = 10;

onlineTicket.PrintTicketInfo();

onlineTicket.GenerateQRCode();
onlineTicket.SendToEmail("example@mail.ru");

Console.WriteLine(
    $"Цена со скидкой: {onlineTicket.GetPriceWithDiscount()} руб.");

// ------------------------------------------------------
// ПЕРЕГРУЗКА МЕТОДОВ
// ------------------------------------------------------

Console.WriteLine("\n--- Перегрузка методов ---");

TicketHelper helper = new TicketHelper();

helper.PrintTicket("Варвара");
helper.PrintTicket("Варвара", 10);
helper.PrintTicket("Варвара", 10, 5000);

// ------------------------------------------------------
// ПОЛИМОРФИЗМ
// ------------------------------------------------------

Console.WriteLine("\n--- Полиморфизм ---");

List<Ticket> tickets = new List<Ticket>();

tickets.Add(concert);
tickets.Add(playTicket);
tickets.Add(conferenceTicket);
tickets.Add(vipTicket);
tickets.Add(onlineTicket);

foreach (Ticket ticket in tickets)
{
    ticket.PrintTicketInfo();
}

// ------------------------------------------------------
// TICKETSYSTEM
// ------------------------------------------------------

Console.WriteLine("\n--- Работа TicketSystem ---");

TicketSystem system = new TicketSystem();

system.SystemName = "My Ticket System";
system.City = "Тюмень";

system.AddTicket(concertTicket);
system.AddTicket(playTicket);
system.AddTicket(conferenceTicket);
system.AddTicket(vipTicket);
system.AddTicket(onlineTicket);

system.ShowSystemInfo();

system.ShowAllTickets();

// ------------------------------------------------------
// БРОНИРОВАНИЕ
// ------------------------------------------------------

Console.WriteLine("\n--- Бронирование ---");

system.ReserveTicket(1);
system.ReserveTicket(4);

// Повторное бронирование
system.ReserveTicket(1);

// ------------------------------------------------------
// ДОП МЕТОДЫ TICKET
// ------------------------------------------------------

Console.WriteLine("\n--- Дополнительные методы ---");

concert.ChangeSeat(10, 25);
concert.SetBuyer("Варвара");
concert.CancelReservation();

// ------------------------------------------------------
// ОДРОБНАЯ ИНФОРМАЦИЯ
// ------------------------------------------------------

Console.WriteLine("\n--- Подробная информация ---");

Console.WriteLine(playTicket.GetTicketDetails());

Console.WriteLine(
    $"Всего билетов в системе: {system.GetTicketCount()}");
```
