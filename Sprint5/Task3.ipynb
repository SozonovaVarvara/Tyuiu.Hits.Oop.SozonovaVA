Требования к базовому классу Ticket:

```
using System;
using System.Collections.Generic;

// 1. ДЕЛЕГАТЫ

// Делегат для выполнения операций с билетами
public delegate void TicketAction(Ticket ticket);

// Делегат для фильтрации билетов
public delegate bool TicketFilter(Ticket ticket);

// 2. ИНТЕРФЕЙСЫ

public interface ITicket
{
    void ShowInfo();
    string GetDetails();
}

public interface IReservation
{
    void Reserve();
    void Cancel();
}

public interface IEventInfo
{
    string GetEventName();
}

public interface INotificationService
{
    void Notify(string message);
}

public interface ITicketRepository
{
    void Add(Ticket ticket);
    Ticket FindById(int id);
    List<Ticket> GetAll();
    void Remove(int id);
}

// 3. КЛАСС ДЛЯ СОБЫТИЙ

public class TicketEventArgs : EventArgs
{
    public string Message { get; set; }

    public TicketEventArgs(string message)
    {
        Message = message;
    }
}

// 4. БАЗОВЫЙ КЛАСС TICKET

public class Ticket : ITicket, IReservation
{
    // Основные атрибуты
    public int TicketId { get; set; }
    public int RowNumber { get; set; }
    public int SeatNumber { get; set; }
    public bool IsReserved { get; private set; }

    // Дополнительные атрибуты
    public decimal Price { get; set; }
    public string EventDate { get; set; }
    public string Category { get; set; }

    // Конструктор
    public Ticket(int ticketId, int rowNumber, int seatNumber)
    {
        TicketId = ticketId;
        RowNumber = rowNumber;
        SeatNumber = seatNumber;
        IsReserved = false;

        Price = 1000;
        EventDate = "Не указана";
        Category = "Обычный";
    }

    // Вывод информации
    public virtual void PrintTicketInfo()
    {
        Console.WriteLine(
            $"Билет #{TicketId}: ряд {RowNumber}, место {SeatNumber}");
    }

    // Бронирование
    public virtual void ReserveSeat()
    {
        if (!IsReserved)
        {
            IsReserved = true;
            Console.WriteLine("Место успешно забронировано!");
        }
        else
        {
            Console.WriteLine("Место уже забронировано.");
        }
    }

    // Получение информации
    public virtual string GetTicketDetails()
    {
        return $"ID: {TicketId}, Цена: {Price}, " +
               $"Категория: {Category}, Дата: {EventDate}, " +
               $"Забронирован: {IsReserved}";
    }

    // Изменение места
    public void ChangeSeat(int row, int seat)
    {
        if (!IsReserved)
        {
            RowNumber = row;
            SeatNumber = seat;
            Console.WriteLine("Место изменено.");
        }
    }

    // Изменение цены
    public void ChangePrice(decimal newPrice)
    {
        if (newPrice > 0)
            Price = newPrice;
    }

    // Отмена бронирования
    public virtual void CancelReservation()
    {
        IsReserved = false;
        Console.WriteLine("Бронирование отменено.");
    }

    // Явная реализация ITicket
    void ITicket.ShowInfo()
    {
        PrintTicketInfo();
    }

    string ITicket.GetDetails()
    {
        return GetTicketDetails();
    }

    // Явная реализация IReservation
    void IReservation.Reserve()
    {
        ReserveSeat();
    }

    void IReservation.Cancel()
    {
        CancelReservation();
    }
}

// 5. CONCERTTICKET

public class ConcertTicket : Ticket, IEventInfo
{
    public string ArtistName { get; set; }
    public string MusicGenre { get; set; }
    public string Stage { get; set; }

    public ConcertTicket(
        int id, int row, int seat, string artist)
        : base(id, row, seat)
    {
        ArtistName = artist;
        MusicGenre = "Pop";
        Stage = "Главная сцена";
    }

    public override void PrintTicketInfo()
    {
        Console.WriteLine(
            $"[Концерт] {ArtistName}, ряд {RowNumber}, " +
            $"место {SeatNumber}");
    }

    public override string GetTicketDetails()
    {
        return base.GetTicketDetails() +
               $", Исполнитель: {ArtistName}, Жанр: {MusicGenre}";
    }

    // Изменение исполнителя
    public void ChangeArtist(string artist)
    {
        ArtistName = artist;
    }

    // Изменение жанра
    public void ChangeGenre(string genre)
    {
        MusicGenre = genre;
    }

    // Явная реализация интерфейса
    string IEventInfo.GetEventName()
    {
        return "Концерт: " + ArtistName;
    }
}

// 6. PLAYTICKET

public class PlayTicket : Ticket, IEventInfo
{
    public string PlayTitle { get; set; }
    public string TheaterName { get; set; }
    public string Genre { get; set; }

    public PlayTicket(
        int id, int row, int seat, string title)
        : base(id, row, seat)
    {
        PlayTitle = title;
        TheaterName = "Городской театр";
        Genre = "Драма";
    }

    public override void PrintTicketInfo()
    {
        Console.WriteLine(
            $"[Спектакль] {PlayTitle}, ряд {RowNumber}, " +
            $"место {SeatNumber}");
    }

    public override string GetTicketDetails()
    {
        return base.GetTicketDetails() +
               $", Спектакль: {PlayTitle}, Театр: {TheaterName}";
    }

    // Изменение названия спектакля
    public void ChangeTitle(string title)
    {
        PlayTitle = title;
    }

    // Изменение театра
    public void ChangeTheater(string theater)
    {
        TheaterName = theater;
    }

    // Явная реализация интерфейса
    string IEventInfo.GetEventName()
    {
        return "Спектакль: " + PlayTitle;
    }
}

// 7. CONFERENCETICKET

public class ConferenceTicket : Ticket, IEventInfo
{
    public string ConferenceName { get; set; }
    public string SeatPreference { get; set; }
    public string Organization { get; set; }

    public ConferenceTicket(
        int id, int row, int seat, string name)
        : base(id, row, seat)
    {
        ConferenceName = name;
        SeatPreference = "Без предпочтений";
        Organization = "Не указана";
    }

    public override void PrintTicketInfo()
    {
        Console.WriteLine(
            $"[Конференция] {ConferenceName}, " +
            $"ряд {RowNumber}, место {SeatNumber}");
    }

    public override string GetTicketDetails()
    {
        return base.GetTicketDetails() +
               $", Конференция: {ConferenceName}, " +
               $"Предпочтение: {SeatPreference}";
    }

    // Изменение предпочтения места
    public void SetSeatPreference(string preference)
    {
        SeatPreference = preference;
    }

    // Изменение организации
    public void ChangeOrganization(string organization)
    {
        Organization = organization;
    }

    // Переопределение бронирования
    public override void ReserveSeat()
    {
        Console.WriteLine("Предпочтение: " + SeatPreference);
        base.ReserveSeat();
    }

    // Явная реализация интерфейса
    string IEventInfo.GetEventName()
    {
        return "Конференция: " + ConferenceName;
    }
}

// 8. КОЛЛЕКЦИИ И ХРАНИЛИЩЕ БИЛЕТОВ

public class TicketRepository : ITicketRepository
{
    // Коллекция List для хранения всех билетов
    private List<Ticket> tickets = new List<Ticket>();

    // Коллекция Dictionary для быстрого поиска по ID
    private Dictionary<int, Ticket> ticketDictionary =
        new Dictionary<int, Ticket>();

    // Добавление билета
    public void Add(Ticket ticket)
    {
        if (!ticketDictionary.ContainsKey(ticket.TicketId))
        {
            tickets.Add(ticket);
            ticketDictionary.Add(ticket.TicketId, ticket);
        }
        else
        {
            Console.WriteLine("Такой билет уже существует.");
        }
    }

    // Поиск билета по ID
    public Ticket FindById(int id)
    {
        if (ticketDictionary.ContainsKey(id))
            return ticketDictionary[id];

        return null;
    }

    // Получение всех билетов
    public List<Ticket> GetAll()
    {
        return new List<Ticket>(tickets);
    }

    // Удаление билета
    public void Remove(int id)
    {
        Ticket ticket = FindById(id);

        if (ticket != null)
        {
            tickets.Remove(ticket);
            ticketDictionary.Remove(id);
            Console.WriteLine("Билет удален.");
        }
        else
        {
            Console.WriteLine("Билет не найден.");
        }
    }
}

// 9. СЛУЖБА УВЕДОМЛЕНИЙ

public class ConsoleNotificationService : INotificationService
{
    public void Notify(string message)
    {
        Console.WriteLine("[Уведомление] " + message);
    }
}

// 10. СИСТЕМА УПРАВЛЕНИЯ БИЛЕТАМИ

public class TicketSystem
{
    private ITicketRepository repository;
    private INotificationService notification;

    // СОБЫТИЕ
    public event EventHandler<TicketEventArgs> TicketChanged;

    // Конструктор с внедрением зависимостей
    public TicketSystem(
        ITicketRepository repository,
        INotificationService notification)
    {
        this.repository = repository;
        this.notification = notification;
    }

    // Метод вызова события
    protected virtual void OnTicketChanged(string message)
    {
        TicketChanged?.Invoke(
            this, new TicketEventArgs(message));
    }

    // Добавление билета
    public void AddTicket(Ticket ticket)
    {
        if (repository.FindById(ticket.TicketId) != null)
        {
            notification.Notify("Такой билет уже существует.");
            return;
        }

        repository.Add(ticket);

        OnTicketChanged(
            $"Добавлен билет #{ticket.TicketId}");
    }

    // Вывод всех билетов
    public void ShowAllTickets()
    {
        Console.WriteLine("\n--- Все билеты ---");

        foreach (Ticket ticket in repository.GetAll())
        {
            ticket.PrintTicketInfo();
        }
    }

    // Бронирование билета
    public void ReserveTicket(int id)
    {
        Ticket ticket = repository.FindById(id);

        if (ticket != null)
        {
            bool wasReserved = ticket.IsReserved;

            ticket.ReserveSeat();

            if (!wasReserved && ticket.IsReserved)
            {
                OnTicketChanged(
                    $"Билет #{id} забронирован");
            }
        }
        else
        {
            notification.Notify("Билет не найден.");
        }
    }

    // Отмена бронирования
    public void CancelTicket(int id)
    {
        Ticket ticket = repository.FindById(id);

        if (ticket != null && ticket.IsReserved)
        {
            IReservation reservation = ticket;
            reservation.Cancel();

            OnTicketChanged(
                $"Бронирование билета #{id} отменено");
        }
    }

    // Удаление билета
    public void DeleteTicket(int id)
    {
        if (repository.FindById(id) != null)
        {
            repository.Remove(id);

            OnTicketChanged(
                $"Билет #{id} удален");
        }
    }

    // ФИЛЬТРАЦИЯ С ПОМОЩЬЮ ДЕЛЕГАТА
    public void FilterTickets(TicketFilter filter)
    {
        Console.WriteLine("\n--- Результат фильтрации ---");

        foreach (Ticket ticket in repository.GetAll())
        {
            if (filter(ticket))
            {
                ticket.PrintTicketInfo();
            }
        }
    }

    // ВЫПОЛНЕНИЕ ОПЕРАЦИИ С ПОМОЩЬЮ ДЕЛЕГАТА
    public void ProcessTickets(TicketAction action)
    {
        foreach (Ticket ticket in repository.GetAll())
        {
            action(ticket);
        }
    }
}

// 11. ГЛАВНАЯ ПРОГРАММА

public class Program
{
    public static void Main()
    {
        // Создание билетов
        ConcertTicket concert =
            new ConcertTicket(1, 5, 34, "Kanye West");

        PlayTicket play =
            new PlayTicket(2, 2, 13, "Лебединое озеро");

        ConferenceTicket conference =
            new ConferenceTicket(4, 7, 15, "TechConf 2026");

        conference.SetSeatPreference("Ближе к сцене");

        // Создание зависимостей
        ITicketRepository repository =
            new TicketRepository();

        INotificationService notification =
            new ConsoleNotificationService();

        // Создание системы
        TicketSystem system =
            new TicketSystem(repository, notification);

        // ПОДПИСКА НА СОБЫТИЕ
        system.TicketChanged += (sender, e) =>
        {
            notification.Notify(e.Message);
        };

        // Добавление билетов
        Console.WriteLine("--- Добавление билетов ---");

        system.AddTicket(concert);
        system.AddTicket(play);
        system.AddTicket(conference);

        // Вывод всех билетов
        system.ShowAllTickets();

        // Бронирование
        Console.WriteLine("\n--- Бронирование ---");

        system.ReserveTicket(4);
        system.ReserveTicket(1);

        // Отмена бронирования
        Console.WriteLine("\n--- Отмена бронирования ---");

        system.CancelTicket(1);

        // РАБОТА С ДЕЛЕГАТОМ
        Console.WriteLine("\n--- Фильтрация по цене ---");

        TicketFilter expensiveTickets =
            ticket => ticket.Price >= 1000;

        system.FilterTickets(expensiveTickets);

        // Изменение цены всех билетов через делегат
        Console.WriteLine("\n--- Изменение цен ---");

        TicketAction changePrice = ticket =>
        {
            ticket.ChangePrice(ticket.Price + 500);
        };

        system.ProcessTickets(changePrice);

        // Повторный вывод информации
        system.ShowAllTickets();

        // Удаление билета
        Console.WriteLine("\n--- Удаление билета ---");

        system.DeleteTicket(2);

        // Проверка явной реализации интерфейса
        Console.WriteLine("\n--- Работа с интерфейсом ---");

        ITicket ticketInterface = concert;

        ticketInterface.ShowInfo();

        Console.WriteLine(ticketInterface.GetDetails());

        IEventInfo eventInfo = conference;

        Console.WriteLine(eventInfo.GetEventName());
    }
}
Program.Main();
```
