[Создать базовый класс Ticket в C#, который будет представлять билеты на различные мероприятия. На основе этого класса разработать 2-3 производных класса, демонстрирующих принципы наследования и полиморфизма. В каждом из классов должны быть реализованы новые атрибуты и методы, а также переопределены некоторые методы базового класса для демонстрации полиморфизма.

Требования к базовому классу Ticket:
    • Атрибуты: ID билета (TicketId), Номер ряда (RowNumber), Номер места (SeatNumber). 
    • Методы:  
        o PrintTicketInfo(): метод для печати информации о билете.
        o ReserveSeat(): метод для резервирования места. 
        o GetTicketDetails(): метод для получения деталей билета.

Требования к производным классам:
1) БилетНаКонцерт (ConcertTicket): Должен содержать дополнительные атрибуты, такие как Имя исполнителя (ArtistName). Метод PrintTicketInfo() должен быть переопределен для включения имени исполнителя в информацию о билете.
2) БилетНаСпектакль (PlayTicket): Должен содержать дополнительные атрибуты, такие как Название спектакля (PlayTitle). Метод GetTicketDetails() должен быть переопределен для отображения названия спектакля вместе с другими деталями билета.
3) БилетНаКонференцию (ConferenceTicket) (если требуется третий класс): Должен содержать дополнительные атрибуты, такие как Название конференции (ConferenceName). Метод ReserveSeat() должен быть переопределен для добавления информации о предпочтениях участника относительно мест.]

#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) и реализуйте полиморфизм с перекрытием и прегегрузкой методов, а также generic классы

```
using System.Collections.Generic;

// БАЗОВЫЙ КЛАСС
public class Ticket
{
    public int TicketId { get; set; }
    public int RowNumber { get; set; }
    public int SeatNumber { get; set; }
    public bool IsReserved { get; private set; }

    // Добавленные атрибуты
    public double Price { get; set; }
    public string Date { get; set; }
    public string Hall { get; set; }
    public string BuyerName { get; set; }

    public Ticket(int ticketId, int rowNumber, int seatNumber)
    {
        TicketId = ticketId;
        RowNumber = rowNumber;
        SeatNumber = seatNumber;
        IsReserved = false;
        Price = 0;
        Date = "";
        Hall = "";
        BuyerName = "";
    }

    public virtual void PrintTicketInfo()
    {
        Console.WriteLine(
            $"Билет #{TicketId}: ряд {RowNumber}, место {SeatNumber}");
    }

    public virtual void ReserveSeat()
    {
        if (!IsReserved)
        {
            IsReserved = true;
            Console.WriteLine(
                $"Место {SeatNumber} в ряду {RowNumber} успешно забронировано!");
        }
        else
        {
            Console.WriteLine("Это место уже забронировано.");
        }
    }

    public virtual string GetTicketDetails()
    {
        return $"Ticket ID: {TicketId}, Row: {RowNumber}, " +
               $"Seat: {SeatNumber}, Reserved: {IsReserved}, " +
               $"Price: {Price}, Date: {Date}, Hall: {Hall}, " +
               $"Buyer: {BuyerName}";
    }

    // Добавленные методы
    public void ChangeSeat(int newRow, int newSeat)
    {
        RowNumber = newRow;
        SeatNumber = newSeat;
        Console.WriteLine(
            $"Новое место: ряд {RowNumber}, место {SeatNumber}");
    }

    public void SetBuyer(string name)
    {
        BuyerName = name;
        Console.WriteLine($"Покупатель: {BuyerName}");
    }

    public void CancelReservation()
    {
        if (IsReserved)
        {
            IsReserved = false;
            Console.WriteLine("Бронирование отменено.");
        }
        else
        {
            Console.WriteLine("Билет не был забронирован.");
        }
    }

    public virtual void ShowPrice()
    {
        Console.WriteLine($"Цена билета: {Price} руб.");
    }

    // Перегрузка метода
    public void ChangeSeat(int newSeat)
    {
        SeatNumber = newSeat;
        Console.WriteLine($"Место изменено на {SeatNumber}");
    }
}


// БИЛЕТ НА КОНЦЕРТ
public class ConcertTicket : Ticket
{
    public string ArtistName { get; set; }

    // Добавленные атрибуты
    public string Genre { get; set; }
    public string Stage { get; set; }
    public bool MeetAndGreet { get; set; }

    public ConcertTicket(
        int ticketId,
        int rowNumber,
        int seatNumber,
        string artistName)
        : base(ticketId, rowNumber, seatNumber)
    {
        ArtistName = artistName;
        Genre = "";
        Stage = "";
        MeetAndGreet = false;
    }

    // Перекрытие метода
    public override void PrintTicketInfo()
    {
        Console.WriteLine(
            $"[КОНЦЕРТ] Билет #{TicketId}: {ArtistName}, " +
            $"ряд {RowNumber}, место {SeatNumber}");
    }

    public override void ShowPrice()
    {
        Console.WriteLine(
            $"Цена билета на концерт {ArtistName}: {Price} руб.");
    }

    // Добавленные методы
    public void ShowArtist()
    {
        Console.WriteLine($"Исполнитель: {ArtistName}");
    }

    public void ChangeStage(string newStage)
    {
        Stage = newStage;
        Console.WriteLine($"Сцена: {Stage}");
    }

    public void AddMeetAndGreet()
    {
        MeetAndGreet = true;
        Console.WriteLine("В билет добавлена встреча с артистом.");
    }

    public void SetGenre(string genre)
    {
        Genre = genre;
        Console.WriteLine($"Жанр: {Genre}");
    }
}


// БИЛЕТ НА СПЕКТАКЛЬ
public class PlayTicket : Ticket
{
    public string PlayTitle { get; set; }

    // Добавленные атрибуты
    public string Director { get; set; }
    public string TheaterName { get; set; }
    public string Language { get; set; }

    public PlayTicket(
        int ticketId,
        int rowNumber,
        int seatNumber,
        string playTitle)
        : base(ticketId, rowNumber, seatNumber)
    {
        PlayTitle = playTitle;
        Director = "";
        TheaterName = "";
        Language = "Русский";
    }

    // Перекрытие метода
    public override string GetTicketDetails()
    {
        return $"[СПЕКТАКЛЬ] {PlayTitle}, " +
               $"ID: {TicketId}, ряд: {RowNumber}, " +
               $"место: {SeatNumber}, цена: {Price} руб.";
    }

    public override void ShowPrice()
    {
        Console.WriteLine(
            $"Цена билета на спектакль: {Price} руб.");
    }

    // Добавленные методы
    public void ShowPlay()
    {
        Console.WriteLine($"Спектакль: {PlayTitle}");
    }

    public void SetDirector(string director)
    {
        Director = director;
        Console.WriteLine($"Режиссер: {Director}");
    }

    public void SetTheater(string theater)
    {
        TheaterName = theater;
        Console.WriteLine($"Театр: {TheaterName}");
    }

    public void ChangeLanguage(string language)
    {
        Language = language;
        Console.WriteLine($"Язык спектакля: {Language}");
    }
}


// БИЛЕТ НА КОНФЕРЕНЦИЮ
public class ConferenceTicket : Ticket
{
    public string ConferenceName { get; set; }
    public string SeatPreference { get; set; }

    // Добавленные атрибуты
    public string Speaker { get; set; }
    public string Topic { get; set; }
    public bool HasCertificate { get; set; }
    public string Company { get; set; }

    public ConferenceTicket(
        int ticketId,
        int rowNumber,
        int seatNumber,
        string conferenceName)
        : base(ticketId, rowNumber, seatNumber)
    {
        ConferenceName = conferenceName;
        SeatPreference = "";
        Speaker = "";
        Topic = "";
        HasCertificate = false;
        Company = "";
    }

    // Перекрытие метода
    public override void ReserveSeat()
    {
        if (!string.IsNullOrEmpty(SeatPreference))
        {
            Console.WriteLine(
                $"Предпочтение места: {SeatPreference}");
        }

        base.ReserveSeat();
    }

    // Добавленные методы
    public void SetSpeaker(string speaker)
    {
        Speaker = speaker;
        Console.WriteLine($"Докладчик: {Speaker}");
    }

    public void SetTopic(string topic)
    {
        Topic = topic;
        Console.WriteLine($"Тема: {Topic}");
    }

    public void AddCertificate()
    {
        HasCertificate = true;
        Console.WriteLine("Сертификат участника добавлен.");
    }

    public void SetCompany(string company)
    {
        Company = company;
        Console.WriteLine($"Компания: {Company}");
    }
}


// GENERIC-КЛАСС
public class TicketCollection<T> where T : Ticket
{
    private List<T> items = new List<T>();

    public void Add(T item)
    {
        items.Add(item);
        Console.WriteLine(
            $"Объект #{item.TicketId} добавлен в коллекцию.");
    }

    public T Get(int index)
    {
        return items[index];
    }

    public int Count()
    {
        return items.Count;
    }

    public void ShowAll()
    {
        Console.WriteLine("\n--- Generic коллекция ---");

        foreach (T item in items)
        {
            item.PrintTicketInfo();
        }
    }
}


// GENERIC-КЛАСС С ДВУМЯ ТИПАМИ
public class TicketStorage<TKey, TValue>
{
    private Dictionary<TKey, TValue> data =
        new Dictionary<TKey, TValue>();

    public void Add(TKey key, TValue value)
    {
        data.Add(key, value);
    }

    public TValue Get(TKey key)
    {
        return data[key];
    }

    public int Count()
    {
        return data.Count;
    }
}


// КЛАСС ДЛЯ ПЕРЕГРУЗКИ
public class TicketHelper
{
    public void PrintTicket(string name)
    {
        Console.WriteLine($"Покупатель: {name}");
    }

    public void PrintTicket(string name, int ticketId)
    {
        Console.WriteLine(
            $"Покупатель: {name}, билет #{ticketId}");
    }

    public void PrintTicket(
        string name,
        int ticketId,
        double price)
    {
        Console.WriteLine(
            $"Покупатель: {name}, билет #{ticketId}, " +
            $"цена: {price} руб.");
    }
}


// ======================================================
// РАБОТА ПРОГРАММЫ
// ======================================================

Console.WriteLine("=================================");
Console.WriteLine("       СИСТЕМА БИЛЕТОВ");
Console.WriteLine("=================================");


// Создание объектов
ConcertTicket concertTicket =
    new ConcertTicket(1, 5, 34, "Kanye West");

PlayTicket playTicket =
    new PlayTicket(2, 2, 13, "Лебединое озеро");

ConferenceTicket conferenceTicket =
    new ConferenceTicket(3, 7, 15, "TechConf 2026");


// Заполнение атрибутов
concertTicket.Price = 5000;
concertTicket.Date = "20.12.2026";
concertTicket.Hall = "Главный зал";
concertTicket.BuyerName = "Варвара";

concertTicket.SetGenre("Pop");
concertTicket.ChangeStage("Main Stage");
concertTicket.AddMeetAndGreet();

playTicket.Price = 3500;
playTicket.Date = "25.12.2026";
playTicket.Hall = "Большой театр";

playTicket.SetDirector("Алексей Иванов");
playTicket.SetTheater("Большой театр");
playTicket.ChangeLanguage("Русский");

conferenceTicket.Price = 2500;
conferenceTicket.Date = "10.11.2026";
conferenceTicket.Hall = "Конференц-зал";
conferenceTicket.SeatPreference = "Ближе к сцене";

conferenceTicket.SetSpeaker("Иван Петров");
conferenceTicket.SetTopic("Искусственный интеллект");
conferenceTicket.AddCertificate();
conferenceTicket.SetCompany("IT Company");


// ПОЛИМОРФИЗМ С ПЕРЕКРЫТИЕМ
Console.WriteLine("\n--- ПОЛИМОРФИЗМ ---");

List<Ticket> tickets = new List<Ticket>();

tickets.Add(concertTicket);
tickets.Add(playTicket);
tickets.Add(conferenceTicket);

foreach (Ticket ticket in tickets)
{
    ticket.PrintTicketInfo();
    ticket.ShowPrice();
}


// БРОНИРОВАНИЕ
Console.WriteLine("\n--- БРОНИРОВАНИЕ ---");

concertTicket.ReserveSeat();
conferenceTicket.ReserveSeat();


// ПЕРЕГРУЗКА МЕТОДОВ
Console.WriteLine("\n--- ПЕРЕГРУЗКА МЕТОДОВ ---");

TicketHelper helper = new TicketHelper();

helper.PrintTicket("Варвара");
helper.PrintTicket("Варвара", 1);
helper.PrintTicket("Варвара", 1, 5000);


// Перегрузка ChangeSeat
Console.WriteLine("\n--- ПЕРЕГРУЗКА ChangeSeat ---");

concertTicket.ChangeSeat(10, 25);
playTicket.ChangeSeat(20);


// GENERIC-КЛАСС
Console.WriteLine("\n--- GENERIC КЛАСС ---");

TicketCollection<Ticket> collection =
    new TicketCollection<Ticket>();

collection.Add(concertTicket);
collection.Add(playTicket);
collection.Add(conferenceTicket);

collection.ShowAll();

Console.WriteLine(
    $"Количество билетов: {collection.Count()}");


// GENERIC ДЛЯ ПРОИЗВОДНОГО КЛАССА
Console.WriteLine("\n--- GENERIC ДЛЯ CONCERTTICKET ---");

TicketCollection<ConcertTicket> concertCollection =
    new TicketCollection<ConcertTicket>();

concertCollection.Add(concertTicket);
concertCollection.ShowAll();


// GENERIC С ДВУМЯ ТИПАМИ
Console.WriteLine("\n--- GENERIC STORAGE ---");

TicketStorage<int, Ticket> storage =
    new TicketStorage<int, Ticket>();

storage.Add(1, concertTicket);
storage.Add(2, playTicket);
storage.Add(3, conferenceTicket);

Ticket savedTicket = storage.Get(1);

Console.WriteLine(
    $"Получен билет #{savedTicket.TicketId}");

Console.WriteLine(
    $"Всего элементов в Storage: {storage.Count()}");


// ПОДРОБНАЯ ИНФОРМАЦИЯ
Console.WriteLine("\n--- ПОДРОБНАЯ ИНФОРМАЦИЯ ---");

Console.WriteLine(
    playTicket.GetTicketDetails());
```
