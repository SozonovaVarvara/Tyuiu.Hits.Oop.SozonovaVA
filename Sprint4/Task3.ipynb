Требования к базовому классу Ticket:

    • Методы:

```
using System;
using System.Collections.Generic;

// 1. Создание интерфейсов

// Интерфейс для получения информации о билете
public interface ITicket
{
    void ShowInfo();
    string GetDetails();
}

// Интерфейс для бронирования
public interface IReservation
{
    void Reserve();
    void Cancel();
}

// Интерфейс для получения названия мероприятия
public interface IEventInfo
{
    string GetEventName();
}

// Интерфейс для отправки уведомлений
public interface INotificationService
{
    void Notify(string message);
}

// Интерфейс для хранения билетов
public interface ITicketRepository
{
    void Add(Ticket ticket);
    Ticket FindById(int id);
    List<Ticket> GetAll();
}

// 2. Базовый класс TICKET
public class Ticket : ITicket, IReservation
{
    // Основные свойства
    public int TicketId { get; set; }
    public int RowNumber { get; set; }
    public int SeatNumber { get; set; }
    public bool IsReserved { get; private set; }

    // Дополнительные свойства
    public decimal Price { get; set; }
    public string EventDate { get; set; }
    public string Category { get; set; }

    // Конструктор
    public Ticket(int ticketId, int rowNumber, int seatNumber)
    {
        TicketId = ticketId;
        RowNumber = rowNumber;
        SeatNumber = seatNumber;
        IsReserved = false;

        Price = 1000;
        EventDate = "Не указана";
        Category = "Обычный";
    }

    // Обычный метод вывода информации
    public virtual void PrintTicketInfo()
    {
        Console.WriteLine(
            $"Билет #{TicketId}: ряд {RowNumber}, место {SeatNumber}");
    }

    // Бронь
    public virtual void ReserveSeat()
    {
        if (!IsReserved)
        {
            IsReserved = true;
            Console.WriteLine("Место успешно забранировано!");
        }
        else
        {
            Console.WriteLine("Место уже забранировано.");
        }
    }

    // Получение информации
    public virtual string GetTicketDetails()
    {
        return $"ID: {TicketId}, Цена: {Price}, " +
               $"Категория: {Category}, Дата: {EventDate}";
    }

    // Доп метод изменения места
    public void ChangeSeat(int row, int seat)
    {
        if (!IsReserved)
        {
            RowNumber = row;
            SeatNumber = seat;
            Console.WriteLine("Место изменено.");
        }
        else
        {
            Console.WriteLine("Нельзя изменить забронированное место.");
        }
    }

    // Доп метод изменения цены
    public void ChangePrice(decimal newPrice)
    {
        if (newPrice > 0)
            Price = newPrice;
    }

    // Отмена брони
    public virtual void CancelReservation()
    {
        IsReserved = false;
        Console.WriteLine("Бронирование отменено.");
    }

    // ЯВНАЯ РЕАЛИЗАЦИЯ ИНТЕРФЕЙСА
    void ITicket.ShowInfo()
    {
        PrintTicketInfo();
    }

    string ITicket.GetDetails()
    {
        return GetTicketDetails();
    }

    // ЯВНАЯ РЕАЛИЗАЦИЯ ИНТЕРФЕЙСА
    void IReservation.Reserve()
    {
        ReserveSeat();
    }

    void IReservation.Cancel()
    {
        CancelReservation();
    }
}

// 3. Производный класс CONCERTTICKET
public class ConcertTicket : Ticket, IEventInfo
{
    public string ArtistName { get; set; }

    // Доп свойства
    public string MusicGenre { get; set; }
    public string Stage { get; set; }

    public ConcertTicket(
        int id, int row, int seat, string artist)
        : base(id, row, seat)
    {
        ArtistName = artist;
        MusicGenre = "Pop";
        Stage = "Главная сцена";
    }

    public override void PrintTicketInfo()
    {
        Console.WriteLine(
            $"[Концерт] {ArtistName}, ряд {RowNumber}, место {SeatNumber}");
    }

    public override string GetTicketDetails()
    {
        return base.GetTicketDetails() +
               $", Исполнитель: {ArtistName}, Жанр: {MusicGenre}";
    }

    // Доп метод
    public void ChangeArtist(string artist)
    {
        ArtistName = artist;
    }

    // ЯВНАЯ РЕАЛИЗАЦИЯ IEventInfo
    string IEventInfo.GetEventName()
    {
        return "Концерт: " + ArtistName;
    }
}

// 4. Производный класс PLAYTICKET
public class PlayTicket : Ticket, IEventInfo
{
    public string PlayTitle { get; set; }

    // Доп свойства
    public string TheaterName { get; set; }
    public string Genre { get; set; }

    public PlayTicket(
        int id, int row, int seat, string title)
        : base(id, row, seat)
    {
        PlayTitle = title;
        TheaterName = "Городской театр";
        Genre = "Драма";
    }

    public override void PrintTicketInfo()
    {
        Console.WriteLine(
            $"[Спектакль] {PlayTitle}, ряд {RowNumber}, место {SeatNumber}");
    }

    public override string GetTicketDetails()
    {
        return base.GetTicketDetails() +
               $", Спектакль: {PlayTitle}, Театр: {TheaterName}";
    }

    // Доп метол
    public void ChangeTheater(string theater)
    {
        TheaterName = theater;
    }

    // ЯВНАЯ РЕАЛИЗАЦИЯ IEventInfo
    string IEventInfo.GetEventName()
    {
        return "Спектакль: " + PlayTitle;
    }
}

// 5. Производный класс CONFERENCETICKET
public class ConferenceTicket : Ticket, IEventInfo
{
    public string ConferenceName { get; set; }

    // Доп свойства
    public string SeatPreference { get; set; }
    public string Organization { get; set; }

    public ConferenceTicket(
        int id, int row, int seat, string name)
        : base(id, row, seat)
    {
        ConferenceName = name;
        SeatPreference = "Без предпочтений";
        Organization = "Не указана";
    }

    public override void PrintTicketInfo()
    {
        Console.WriteLine(
            $"[Конференция] {ConferenceName}, " +
            $"ряд {RowNumber}, место {SeatNumber}");
    }

    public override string GetTicketDetails()
    {
        return base.GetTicketDetails() +
               $", Конференция: {ConferenceName}, " +
               $"Предпочтение: {SeatPreference}";
    }

    // Доп метод
    public void SetSeatPreference(string preference)
    {
        SeatPreference = preference;
    }

    // ЯВНАЯ РЕАЛИЗАЦИЯ IEventInfo
    string IEventInfo.GetEventName()
    {
        return "Конференция: " + ConferenceName;
    }

    // Переопределение бронирования
    public override void ReserveSeat()
    {
        Console.WriteLine("Предпочтение: " + SeatPreference);
        base.ReserveSeat();
    }
}
// 6. Реализация хранилища билетов

public class TicketRepository : ITicketRepository
{
    private List<Ticket> tickets = new List<Ticket>();

    public void Add(Ticket ticket)
    {
        tickets.Add(ticket);
    }

    public Ticket FindById(int id)
    {
        foreach (Ticket ticket in tickets)
        {
            if (ticket.TicketId == id)
                return ticket;
        }

        return null;
    }

    public List<Ticket> GetAll()
    {
        return tickets;
    }
}

// 7. Реализация уведомлений
public class ConsoleNotificationService : INotificationService
{
    public void Notify(string message)
    {
        Console.WriteLine("[Уведомление] " + message);
    }
}

// 8. Система управления билетами
public class TicketSystem
{
    private ITicketRepository repository;
    private INotificationService notification;

    // Внедрение зависимостей через конструктор
    public TicketSystem(
        ITicketRepository repository,
        INotificationService notification)
    {
        this.repository = repository;
        this.notification = notification;
    }

    // Добавление билета
    public void AddTicket(Ticket ticket)
    {
        repository.Add(ticket);

        notification.Notify(
            $"Билет #{ticket.TicketId} добавлен в систему.");
    }

    // Вывод всех билетов
    public void ShowAllTickets()
    {
        foreach (Ticket ticket in repository.GetAll())
        {
            ticket.PrintTicketInfo();
        }
    }

    // Бронирование билета
    public void ReserveTicket(int id)
    {
        Ticket ticket = repository.FindById(id);

        if (ticket != null)
        {
            ticket.ReserveSeat();

            notification.Notify(
                $"Обработа бронирования билета #{id} завершена.");
        }
        else
        {
            notification.Notify("Билет не найден.");
        }
    }
}

public class Program
{
    public static void Main()
    {
        // Объекты билетов
        ConcertTicket concert =
            new ConcertTicket(1, 5, 34, "Kanye West");

        PlayTicket play =
            new PlayTicket(2, 2, 13, "Лебединое озеро");

        ConferenceTicket conference =
            new ConferenceTicket(4, 7, 15, "TechConf 2026");

        conference.SetSeatPreference("Ближе к сцене");

        // Настройка зависимости
        ITicketRepository repository = new TicketRepository();

        INotificationService notification =
            new ConsoleNotificationService();

        // Передаем зависимости в систему
        TicketSystem system =
            new TicketSystem(repository, notification);

        // Добавляем билеты
        system.AddTicket(concert);
        system.AddTicket(play);
        system.AddTicket(conference);

        // Вывод инфы
        Console.WriteLine("\n--- Все билеты ---");
        system.ShowAllTickets();

        // Бронь
        Console.WriteLine("\n--- Бронирование ---");
        system.ReserveTicket(4);

        // Явное обращение к интерфейсам
        Console.WriteLine("\n--- Работа с интерфейсами ---");

        ITicket ticketInterface = concert;
        ticketInterface.ShowInfo();
        Console.WriteLine(ticketInterface.GetDetails());

        IReservation reservationInterface = play;
        reservationInterface.Reserve();
        reservationInterface.Cancel();

        // Вызов интерфейса конкретного мероприятия
        IEventInfo eventInfo = conference;
        Console.WriteLine(eventInfo.GetEventName());
    }
}
//Запуск проги
Program.Main();
```
