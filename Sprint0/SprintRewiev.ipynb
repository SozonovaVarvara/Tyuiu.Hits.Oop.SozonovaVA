```
using System;

// Родительский класс для всех растений
class Plant
{
    // Общие свойства для всех растений
    public string Name { get; set; }       // Название растения
    public string Type { get; set; }       // Тип растения
    public float Height { get; set; }      // Высота растения в см

    // Конструктор класса Plant
    public Plant(string name, string type, float height)
    {
        Name = name;
        Type = type;
        Height = height;
    }

    // virtual позволяет дочерним классам переопределить этот метод
    public virtual void Цветение()
    {
        Console.WriteLine("Растение " + Name + " цветет.");
    }

    // Метод увеличивает высоту растения
    public void Расти(float рост)
    {
        Height += рост;

        Console.WriteLine(Name + " выросло на " + рост +
                          " см. Новая высота: " + Height + " см.");
    }
}

// Класс Flower наследуется от Plant
class Flower : Plant
{
    // Дополнительные свойства цветка
    public string Цвет { get; set; }
    public int КоличествоЛепестков { get; set; }

    // Конструктор Flower
    // base передает общие свойства в конструктор Plant
    public Flower(string name, float height, string цвет, int количествоЛепестков)
        : base(name, "Цветок", height)
    {
        Цвет = цвет;
        КоличествоЛепестков = количествоЛепестков;
    }

    // Переопределяем метод Цветение родительского класса
    public override void Цветение()
    {
        Console.WriteLine("Цветок " + Name +
                          " распускается и радует глаз!");
    }

    // Собственный метод класса Flower
    public void Увядать()
    {
        Console.WriteLine("Цветок " + Name + " начинает увядать.");
    }
}

// Класс Tree наследуется от Plant
class Tree : Plant
{
    // Дополнительные свойства дерева
    public int Возраст { get; set; }
    public string ТипЛистьев { get; set; }

    // Конструктор Tree
    public Tree(string name, float height, int возраст, string типЛистьев)
        : base(name, "Дерево", height)
    {
        Возраст = возраст;
        ТипЛистьев = типЛистьев;
    }

    // Переопределяем метод Цветение для дерева
    public override void Цветение()
    {
        Console.WriteLine("Дерево " + Name +
                          " покрывается новыми листьями!");
    }

    // Собственный метод дерева
    // Увеличивает возраст дерева
    public void Вырасти(int годы)
    {
        Возраст += годы;

        Console.WriteLine("Дерево " + Name +
                          " выросло на " + годы +
                          " год(а). Теперь ему " + Возраст + " лет.");
    }
}

// Создаем объект цветка
Flower flower = new Flower(
    "Роза",       // название
    40,           // высота
    "Красный",    // цвет
    25            // количество лепестков
);

// Создаем объект дерева
Tree tree = new Tree(
    "Береза",     // название
    300,          // высота
    10,           // возраст
    "Лиственные"  // тип листьев
);

// Работа с цветком
Console.WriteLine("Цветок:");

flower.Цветение();   // Вызывается метод Flower
flower.Расти(5);     // Увеличиваем высоту цветка
flower.Увядать();    // Вызываем собственный метод цветка

Console.WriteLine();

// Работа с деревом
Console.WriteLine("Дерево:");

tree.Цветение();     // Вызывается метод Tree
tree.Расти(20);      // Увеличиваем высоту дерева
tree.Вырасти(2);     // Увеличиваем возраст дерева

Console.WriteLine();

// Демонстрация полиморфизма
Console.WriteLine("Полиморфизм:");

// Тип переменных Plant,
// но внутри находятся объекты Flower и Tree
Plant plant1 = flower;
Plant plant2 = tree;

// Из-за полиморфизма вызывается
// переопределенный метод Flower
plant1.Цветение();

// Здесь вызывается переопределенный метод Tree
plant2.Цветение();
```
